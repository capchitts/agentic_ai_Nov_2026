## `range()`

`range()` represents a sequence of integers. It can be called in three forms:

| Form | Values generated |
| --- | --- |
| `range(stop)` | Starts at 0 and stops before `stop`. |
| `range(start, stop)` | Starts at `start`, increases by 1, and stops before `stop`. |
| `range(start, stop, step)` | Starts at `start`, changes by `step`, and stops before `stop`. |

The stop value is always excluded. A negative step counts backward; a zero step is invalid.

````python
# Let's first check the type of the range() function.

type(range(5))
````

### One-Argument Form

With `range(stop)`, Python starts at `0` and stops before `stop`. For example, `range(5)` produces `0, 1, 2, 3, 4`.

````python
range(5) # It should return a lower and an upper bound value.
````

`range(5)` produces the integers 0 through 4. In general, the start is included and the stop is excluded.

````python
# The below way takes two arguments: start and stop.
# The sequence will start from the value given at start, increment by 1 and stop at one less than the value given.

range(0, 5)
````

### Start and Stop Values

In `range(0, 5)`, `0` is included and `5` is excluded, so the sequence is `0, 1, 2, 3, 4`.

### What Does `range()` Return?

In Python 3, `range()` returns a compact range object, not a list. It generates values as they are requested, which is efficient for large ranges. Convert it with `list(range(...))` when you specifically need a list.

## `for` Loops

A `for` loop runs its body once for each value produced by an iterable, such as a list, string, tuple, or `range()`.

```python
for item in items:
    print(item)
```

Read this as: “Take each value from `items`, assign it to `item`, and run the indented body.”

### Read the Loop Structure

```text
for variable in iterable:
    loop body
```

- `variable` receives the current value on each pass.
- `iterable` supplies the values.
- The indented body runs once per value.

A `for` loop is useful when you want to process each value in an iterable. The iterable determines when the loop finishes.

````python
# Lets create a list with name country & try to print all the country names by using for loop

country = ['India', 'UAE', 'United Kingdom']
for i in country:
    print(i)
````

In the previous example, the loop visits each language in order. On every pass, `i` receives the current language and the loop body prints it.

# Continue Statement in python

The continue statement is used to **skip the current iteration of the loop and the control flow of the program goes to the next iteration.**

**Reverse the situation**


The continue is a **loop control statement.**

Using the continue statement, we can stop the **current iteration of the loop based on some condition(s), and control again comes to the beginning of the loop with the next iteration.**

The continue statement can be **used without condition.** But generally, within the for loop in python, we find its condition application to skip some statements in certain iterations.

Let’s see a Python code example below to understand it better. In this, we iterate over a list of different country names and print them. But we want that if the name **‘China’ is encountered, then the name is not printed.**

For this purpose, we have a condition accompanied by a continue statement that skips the print statement and starts the next iteration of the loop.

````python
# Another way of Print out only odd numbers - 1,3,5,7,9

for i in range(10):
    # Check if i is even
    if i % 2 == 0:
        continue
    print(i)
````

````python
# Program to find whether a value is present or not in a sequence

numbers = [5, 9, 10, 7, 13, 3, 20, 11]

# Search Term

item = 15
for i in numbers:
    if i == item:
        print("Item Found.")
        break
else:
    print("Item Not Found!")

````

# Break statement with for loop


The break statement is used to **exit the for loop prematurely,** as it is **loop control statement.**

Basically it’s used to **break the for loop when a specific condition is met** and we stop the **further execution of the for loop usually based on some condition(s).**

Using the break statement, we can **stop the loop or take out the execution control from the loop before its completion.**

Let’s say we have a list of numbers and we want to check if a number is present or not. We can iterate over the list of numbers and if the number is found, break out of the loop because we don’t need to keep iterating over the remaining elements.

The for loop in Python is very similar to other programming languages.

We can use **break and continue statements with for loop to alter the execution.** However, in Python, we can have optional else block in for loop too and in this case we will demonstrate the same.



````python
nums = [1, 2, 3, 4, 5, 6]

n = 2

found = False
for num in nums:
    if n == num:
        found = True
        break

print(f'List contains {n}: {found}')

# Output
# List contains 2: True
````

````python
# Break the loop at 'blue'

colors = ['red', 'green', 'blue', 'yellow']
for x in colors:
    if x == 'blue':
        break
    print(x)
````

````python
# Skip 'blue'

colors = ['red', 'green', 'blue', 'yellow']
for x in colors:
    if x == 'blue':
        continue
    print(x)
````

````python
# Check the condition

target = 3
for number in range(10):
    print(number)
    if number == target:
        break
````

**In a nested loop, the break statement exits the inner loop but continues running the outer loop. So if you call the break statement in the inner loop of a nested loop construct, it doesn’t stop the entire loop but just the inner one.**

## enumerate() function

It adds the **counter to iterable, which starts from zero by default.**

The enumerate() function in Python provides us with a convenient way to **access both the index of an item and the item itself when we’re looping over a list, or any iterable object**

**enumerate(<iterable>, start=0)**

<iterable>: This is a required parameter and can be any Python iterable, such as a list or tuple.

start: This second parameter is an optional parameter that specifies the starting value of the counter. By default, it is set to 0.

````python
# Simple enumerate

fruits = ['apple', 'banana', 'cherry']

names_with_index = enumerate(fruits)

(tuple(names_with_index))
````

## While Loop

With the while loop we can execute a set of statements as long as a condition is true.

**Indefinite Iteration statement** as it will keep on repeatedly executing as long as a while condition is true, and it will stop repeating only if the condition becomes false.

A "while" loop is used when the **number of iterations is unknown.** The statement repeats itself **till the boolean value becomes false.** In a while loop, the **condition is tested at the start,** also known as the **pre-test loop.**

**Syntax:**

while condition:

    # body of while loop

````python
# Count till Count < 5 condition is satisfied

count = 0

while count < 5:
    print(count)
    count += 1  # This is the same as count = count + 1
````

````python
# print n till n-=1

n = 5
while n > 0:
    n -= 1
    print(n)
````

````python
# Printing a string 10 times

count = 0;
while count < 10:
    print("My name is Tarun")
    count += 1
````

## FAQs

**Q: What is the while loop in Python?**

A: The while loop in Python is a control flow statement that allows you to repeatedly execute a block of code as long as a specified condition is true. It provides a way to create iterative processes in your program.

**Q: What is the difference between a while loop and a for loop in Python?**

A: The main difference between a while loop and a for loop is the way they control the flow of execution. A while loop repeatedly executes a block of code as long as a condition is true, whereas a for loop iterates over a sequence (such as a list or a range) and executes the block of code for each element in the sequence.

**Q: How to avoid infinite loops with while loops?**

A: To avoid infinite loops while loops, ensure that the condition within the loop will eventually become False. Otherwise, the loop will continue indefinitely. Make sure to include code within the loop that modifies the variables involved in the condition, allowing the condition to become False at some point.

## Nested Loops in Python

**"Nested means one inside another, therefore nested for loop is a for loop inside another for loop. Or we can say we have a for loop inside the body of another for loop."**

Python allows the use of a **loop within the loop.** This is known as **nesting**.

For every single iteration of the **outer loop, the inner loop executes all its iterations.**

For instance, if the outer loop has n iterations and the inner loop has m iterations, then the total number of iterations will be **n x m.** That is for each iteration of the outer loop. The inner loop will execute m times.

**Cross Join in SQL**



### Let’s see an example below that makes this more clear.

In this, we have two lists named names and city. We want to print each element in the list **‘colors’** with all the items in the list **‘fruits’**

first, we will be iterating over the list **‘colors’** using the for a loop. This will be the outer loop. Now for every iteration of the outer loop, the inner loop will be iterating over the list ‘fruits’.
**The total number of iterations, in this case, will be the multiplication of the number of elements in both lists.**

````python
# Lets see the example to  Combine all the colors and fruits together

colors = ['red','green','yellow','black']
fruits = ['mango','banana','apple']

for i in colors:
    for j in fruits:
        print(i,j)
````

````python
# Python program having nested for loops

names = ['Zeeshan', 'Shubham', 'Vikram']

city = ['Bangalore', 'London', 'Paris']

for i in names:
    for j in city:
        print(f'{i} loves {j}.')
````

## Let’s use a nested for loop to print the multiplication table of the first 10 numbers.

````python

for i in range(1,11):
    for j in range(1,11):
        print(i*j,end=" ")
    print()
````


### Let’s take a closer look at how this program works:

The outer loop iterates numbers from 1 to 10 and stores the current number in a temporary variable i. So in the first iteration, i = 1, on the second, i = 2, and so on.

The inner loop also iterates numbers from 1 to 10. The inner loop stores the current iteration number in a variable j.

For every single outer loop iteration, the inner loop will run entirely, that is, 10 times.

So for example, during the first iteration of the outer loop, i = 1 while j goes from 1 to 10. During the second iteration, i = 2, and j goes from 1 to 10 again, and so on.

The result is a multiplication table where each number 1…10 gets multiplied by 1…10.


````python
n = 5;

print("Pattern for n = 5 is,")

# Iterating over numbers in the range 1 to n

for x in range(1,n+1):

    # Executing the inner loop for x times

    for y in range(x):

   # Printing "*" in each execution of the inner loop

        print("*",end=" ")

    print()
````

### Use a nested for loop to find the prime numbers from 2 to 100

````python
print("Prime numbers in the range 2 to 100 are,")

for i in range(2,101):

    isPrime = True

    # Checking if x is a prime number or not

    for y in range(2,i):

        # Checking if any number in the range [2, i-1] completely divides i

        if i%y==0:

        # If the number completely divides i then x is not a prime number

            isPrime = False
            break

    # If isPrime is true print x

    if isPrime:
        print(i,end=" ")
````
