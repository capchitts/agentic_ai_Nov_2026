## Lists: ordered, mutable sequences

A list keeps its items in order, permits duplicate values, and can be changed after creation. Indexing starts at `0`; negative indexes count from the end. A slice such as `items[1:4]` includes index 1 and stops before index 4. A slice creates a new list; methods such as `append()` and `reverse()` change the original list.

As you run the examples below, keep track of whether an operation returns a new value or mutates the list in place.

A tuple supports indexing and slicing like a list, but item assignment is not supported. Use one when the values belong together and should not be reassigned, such as a coordinate pair. A one-item tuple needs a trailing comma: `(42,)`.

## Sets: unique values without positions

A set stores unique hashable values. Adding a value already present leaves the set unchanged. Sets are useful for membership checks and set operations, but they are not indexed sequences: do not rely on their display order.

````python
# sets
s = {3,4,5,6,7,8}   # set literal: values only, no keys
````

````python
type(s)   # confirm it is a set
````

    {"key1": "niti", 234: [23, 45, 656]},

1. Reverse the outer list without changing it.

3. Retrieve `456` from the first nested list.

````python
# ---------- ANSWERS ----------

l = [3,4,5,6,7 , [23,456,67,8,78,78] , [345,56,87,8,98,9] , (234,6657,6) , {"key1" :"niti" , 234:[23,45,656]}]

# index map:  0-4 -> 3,4,5,6,7   |  5 -> list  |  6 -> list  |  7 -> tuple  |  8 -> dict

# 1. reverse a list
print("1a", l[::-1])        # reverse the whole l  -> [{...}, (234,6657,6), [345,...], [23,...], 7, 6, 5, 4, 3]
print("1b", l[5][::-1])     # reverse the inner list -> [78, 78, 8, 67, 456, 23]
# l.reverse()  <- reverses IN PLACE and returns None

# 2. access 234   (it appears twice: inside the tuple, and as a dict key)
print("2a", l[7][0])            # 234   <- from the tuple
print("2b", list(l[8].keys()))  # ['key1', 234]  <- 234 also exists as a dict key

# 3. access 456
print("3", l[5][1])         # 456

# 4. extract only the list collections out of l
print("4", [x for x in l if type(x) == list])
# [[23, 456, 67, 8, 78, 78], [345, 56, 87, 8, 98, 9]]

# 5. extract "sudh"  -> TRICK: "sudh" is NOT in this data, the value is "niti"
print("5a", l[8]["key1"])       # 'niti'
print("5b", l[8].get("sudh"))   # None   <- safe
# l[8]["sudh"]  -> KeyError: 'sudh'

# 6. all keys of the dict element
print("6", list(l[8].keys()))   # ['key1', 234]

# 7. all values of the dict element
print("7a", list(l[8].values()))  # ['niti', [23, 45, 656]]
print("7b", l[8][234][2])         # 656  <- digging inside a value
````

````python
# ================= PART 2: LOOPS + input() on a REAL order record =================
# Indexing gets you ONE value. Loops get you ALL of them.

order = [
    "ORD-7741",
    3,
    499.99,
    ["laptop_stand", "usb_hub", "cable_2m", "usb_hub"],
    [1, 1, 2, 1],
    ("David Miller", "9876543210", "Pune"),
    {
        "payment": {"mode": "CARD", "txn": "TXN99812", "status": "SUCCESS"},
        "address": {"line1": "Flat 402", "city": "Pune", "pin": 411045},
        "coupons": ["FIRST10", "SAVE50"],
        901: ["packed", "shipped", "out_for_delivery"]
    }
]
order
````

````python
# Q1. Print every item in the cart, one per line, numbered for a receipt.

for i, item in enumerate(order[3], start=1):
    print(i, item)

# 1 laptop_stand
# 2 usb_hub
# 3 cable_2m
# 4 usb_hub

# start=1 because a receipt starts at 1, not 0.
# enumerate hands you the counter AND the value, so no manual range(len(...)) needed.
````

````python
# Q2. Print each item with its quantity.
# order[3] and order[4] are two SEPARATE lists, linked only by position.

for item, qty in zip(order[3], order[4]):
    print(item, "x", qty)

# laptop_stand x 1
# usb_hub x 1
# cable_2m x 2
# usb_hub x 1

# GOTCHA: zip stops at the SHORTER list, silently. No error, data just disappears.
print(list(zip([1,2,3,4], ["a","b"])))   # [(1,'a'), (2,'b')]  <- 2 items lost
````

````python
# Q3. usb_hub appears TWICE. Build item -> total quantity, merging the duplicates.

totals = {}
for item, qty in zip(order[3], order[4]):
    totals[item] = totals.get(item, 0) + qty

print(totals)
# {'laptop_stand': 1, 'usb_hub': 2, 'cable_2m': 2}

# .get(key, 0) means "give me the running total, or 0 if this is the first time".
# This is the single most reused loop in real data work: counting, tallying, grouping.
````

````python
# Q4. Loop the metadata dict, print every key with its value.

for key, value in order[6].items():
    print(key, "->", value)

# payment -> {'mode': 'CARD', 'txn': 'TXN99812', 'status': 'SUCCESS'}
# address -> {'line1': 'Flat 402', 'city': 'Pune', 'pin': 411045}
# coupons -> ['FIRST10', 'SAVE50']
# 901 -> ['packed', 'shipped', 'out_for_delivery']

# .items() gives key AND value together. Looping the dict directly gives only KEYS.
for key in order[6]:
    print(key)   # payment / address / coupons / 901
````

````python
# Q5. Walk the record and print the TYPE of each element.

for element in order:
    print(type(element).__name__, ":", element)

# str : ORD-7741
# int : 3
# float : 499.99
# list : ['laptop_stand', 'usb_hub', 'cable_2m', 'usb_hub']
# list : [1, 1, 2, 1]
# tuple : ('David Miller', '9876543210', 'Pune')
# dict : {...}


# Q6. Print only the collections; for lists also print how many things are inside.

for element in order:
    if isinstance(element, list):
        print("list of", len(element))
    elif isinstance(element, (tuple, dict)):
        print(type(element).__name__, "found")

# list of 4
# list of 4
# tuple found
# dict found

# isinstance(x, (list, tuple, dict)) -> the () holds TYPES to check against, not data.
# It means "is x ANY of these three?"
````

````python
# ---- input() section: from here the data stops being ours and becomes the USER's ----
# Q7. Ask for a product name, say whether it is in the cart.

name = input("Enter product name: ").strip().lower()

if name in order[3]:
    print(name, "is in the cart")
else:
    print(name, "not found in this order")

# .strip()  kills the trailing space users ALWAYS type
# .lower()  so "USB_Hub" still matches "usb_hub"
````

````python
# input() returns a string, so catch the expected error before converting it.
qty_text = input("Quantity: ").strip()
try:
    print(qty_text * order[2])
except TypeError as error:
    print(f"Expected TypeError before conversion: {error}")

# Convert once at the input boundary, then use the numeric value.
qty = int(qty_text)
print("Line amount:", qty * order[2])

# For comparison: "3" * 2 repeats the text and produces "33".
print("3" * 2)
````

````python
# Q9. Ask for a metadata field name, print its value WITHOUT crashing if it's missing.

field = input("Which field? (payment / address / coupons): ").strip()

value = order[6].get(field)
if value is None:
    print("No such field. Available:", list(order[6].keys()))
else:
    print(value)

# order[6][field] would raise KeyError. .get() returns None instead.
# Missing data is the NORMAL case with real APIs, not the exception.


# Q10. Keep asking until the user types an actual number.

while True:
    raw = input("Quantity: ").strip()
    if raw.isdigit():
        qty = int(raw)
        break
    print("Numbers only, please.")

print("Ordering", qty, "units")

# Never trust the first thing a user types.
# A while loop around input() is how EVERY real command-line program handles this.
````

````python
# Q11. CAPSTONE - build the whole cart from the user, then print the bill.
# Everything so far in ONE program: while, break, continue, input(), int(),
# the .get() tally, .items() and sum().

cart = {}

while True:
    item = input("Item name (or 'done'): ").strip().lower()
    if item == "done":
        break

    raw = input("Quantity: ").strip()
    if not raw.isdigit():
        print("Bad quantity, skipping that item.")
        continue          # continue = skip THIS round, keep looping
                          # break    = leave the loop entirely
    cart[item] = cart.get(item, 0) + int(raw)

print("\n--- BILL ---")
for item, qty in cart.items():
    print(item, "x", qty)
print("Total units:", sum(cart.values()))
````

````python
# Q12. Ask for a status update, append it to the delivery history, print newest-first.

new_status = input("New status: ").strip().lower()
order[6][901].append(new_status)

for step in order[6][901][::-1]:
    print(step)

# if you typed "delivered":
# delivered
# out_for_delivery
# shipped
# packed


# ---------------- THREE FAILURES TO TELL APART ----------------
# order[99]           -> IndexError : a LIST, that position does not exist
# order[6]["sudh"]    -> KeyError   : a DICT, that key does not exist
# order[0]["sudh"]    -> TypeError  : a STRING, cannot be indexed by a key at all
#
# .get() is a DICT method only. Lists have no .get() -> guard those with len() or try/except.
````

## Collection Checkpoint

| Need | Choose | Remember |
| --- | --- | --- |
| Ordered values that can change | `list` | Indexing, slicing, duplicates allowed |
| Ordered values whose positions should not be reassigned | `tuple` | Tuple items cannot be reassigned; a contained mutable object may still change |
| Unique values and membership | `set` | No positional indexing; order is not a contract |
| Lookup by a label | `dict` | Membership checks keys; keys must be unique and hashable |
