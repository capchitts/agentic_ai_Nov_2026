## 1️⃣ Datatypes in Python

Python has several built-in datatypes. Let's explore the most common ones.

### 🔢 Numeric Types

````python
# Integers (whole numbers)
age = 25
print(f"age = {age}, type = {type(age)}")

# Floats (decimal numbers)
height = 5.9
print(f"height = {height}, type = {type(height)}")

# Complex numbers
z = 3 + 4j
print(f"z = {z}, type = {type(z)}")
````

### 🔤 Strings

````python
# Strings are sequences of characters
name = "Alice"
print(f"name = {name}, type = {type(name)}")

# String operations
print(f"Uppercase: {name.upper()}")
print(f"Length: {len(name)}")
print(f"First char: {name[0]}")
print(f"Slice [1:4]: {name[1:4]}")

# String formatting
score = 95
print(f"{name} scored {score} points!")
````

### ✅ Booleans

````python
# Booleans: True or False
is_student = True
has_pet = False

print(f"is_student = {is_student}, type = {type(is_student)}")
print(f"has_pet = {has_pet}")

# Boolean operations
print(f"AND: {is_student and has_pet}")
print(f"OR: {is_student or has_pet}")
print(f"NOT: {not is_student}")
````

### 📝 Type Conversion

````python
# Converting between types
num_str = "42"
num_int = int(num_str)
num_float = float(num_str)

print(f"String '{num_str}' -> int: {num_int}")
print(f"String '{num_str}' -> float: {num_float}")
print(f"int {num_int} -> str: {str(num_int)}")

# bool conversion
print(f"bool(0) = {bool(0)}")   # False
print(f"bool(1) = {bool(1)}")   # True
print(f"bool('') = {bool('')}") # False
print(f"bool('hi') = {bool('hi')}") # True
````

## 2️⃣ Loops & Conditions

Control the flow of your program with conditional statements and loops.

### 🔀 Conditional Statements (if/elif/else)

````python
# Basic if-else
temperature = 28

if temperature > 30:
    print("It's hot! 🔥")
elif temperature > 20:
    print("It's warm! ☀️")
else:
    print("It's cold! ❄️")

# One-liner (ternary)
weather = "sunny" if temperature > 25 else "cloudy"
print(f"Weather: {weather}")

# Nested conditions
has_umbrella = True
is_raining = True

if is_raining:
    if has_umbrella:
        print("You're prepared! ☂️")
    else:
        print("You'll get wet! 💦")
else:
    print("No rain today! 🌈")
````

### 🔄 For Loops

````python
# Loop through a list
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(f"I like {fruit}")

# Loop with range()
print("\nCounting to 5:")
for i in range(1, 6):
    print(i, end=" ")

# range(start, stop, step)
print("\n\nEven numbers 0-10:")
for i in range(0, 11, 2):
    print(i, end=" ")

# Loop through a string
print("\n\nLetters in 'Python':")
for letter in "Python":
    print(letter, end=" ")

# enumerate() - get index and value
print("\n\nWith index:")
for idx, fruit in enumerate(fruits):
    print(f"{idx}: {fruit}")
````

### 🔄 While Loops

````python
# Basic while loop
count = 0
while count < 5:
    print(f"Count: {count}")
    count += 1

# while with break
print("\n--- Break example ---")
n = 1
while True:
    print(n, end=" ")
    if n >= 5:
        break
    n += 1

# while with continue
print("\n\n--- Continue example ---")
i = 0
while i < 10:
    i += 1
    if i % 2 == 0:
        continue
    print(f"Odd: {i}")
````

### 🎯 Loop Control: break, continue, else

````python
# for-else: else runs if loop wasn't broken
for n in range(2, 10):
    for x in range(2, n):
        if n % x == 0:
            print(f"{n} = {x} * {n//x}")
            break
    else:
        # Loop wasn't broken → n is prime
        print(f"{n} is a prime number")

# List comprehension (Pythonic way!)
squares = [x**2 for x in range(1, 6)]
print(f"\nSquares: {squares}")

# With condition
evens = [x for x in range(10) if x % 2 == 0]
print(f"Evens: {evens}")
````

## 3️⃣ Data Structures: List, Tuple, Set

### 📋 Lists (Ordered, Mutable, Allows Duplicates)

````python
# Creating lists
fruits = ["apple", "banana", "cherry"]
numbers = [10, 20, 30, 40, 50]
mixed = [1, "hello", 3.14, True]

print(f"Fruits: {fruits}")
print(f"Numbers: {numbers}")

# Accessing elements
print(f"\nFirst fruit: {fruits[0]}")
print(f"Last fruit: {fruits[-1]}")
print(f"Slice [1:3]: {fruits[1:3]}")

# Modifying lists
fruits.append("date")
print(f"\nAfter append: {fruits}")

fruits.insert(1, "apricot")
print(f"After insert: {fruits}")

fruits.remove("banana")
print(f"After remove: {fruits}")

popped = fruits.pop()
print(f"Popped: {popped}, List now: {fruits}")

# Other useful methods
numbers.sort(reverse=True)
print(f"\nSorted descending: {numbers}")
print(f"Count of 20: {numbers.count(20)}")
print(f"Index of 30: {numbers.index(30)}")

# List comprehension
cubes = [x**3 for x in range(1, 6)]
print(f"\nCubes: {cubes}")
````

### 📦 Tuples (Ordered, Immutable, Allows Duplicates)

````python
# Creating tuples
coordinates = (10, 20)
person = ("Alice", 25, "Engineer")
single = (42,)  # Note the comma!

print(f"Coordinates: {coordinates}")
print(f"Person: {person}")
print(f"Type of single: {type(single)}")

# Accessing elements
print(f"\nX: {coordinates[0]}, Y: {coordinates[1]}")
print(f"Name: {person[0]}")

# Tuples are immutable (can't change)
# coordinates[0] = 99  # ❌ This would raise TypeError

# Tuple unpacking
x, y = coordinates
name, age, job = person
print(f"\nUnpacked: x={x}, y={y}")
print(f"Unpacked: {name} is {age} years old, works as {job}")

# Tuple methods
t = (1, 2, 3, 2, 2, 4)
print(f"\nCount of 2: {t.count(2)}")
print(f"Index of 3: {t.index(3)}")

# Tuple as dict key (because it's hashable!)
locations = {(0, 0): "origin", (10, 20): "point A"}
print(f"\nLocation at origin: {locations[(0, 0)]}")
````

### 🎯 Sets (Unordered, Mutable, No Duplicates)

````python
# Creating sets
fruits = {"apple", "banana", "cherry"}
numbers = set([1, 2, 2, 3, 3, 3])  # Removes duplicates
empty_set = set()  # NOT {} (that's a dict!)

print(f"Fruits: {fruits}")
print(f"Numbers (no dupes): {numbers}")

# Adding and removing
fruits.add("date")
print(f"\nAfter add: {fruits}")

fruits.discard("banana")
print(f"After discard: {fruits}")

# Set operations
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(f"\nSet A: {a}")
print(f"Set B: {b}")
print(f"Union (A | B): {a | b}")
print(f"Intersection (A & B): {a & b}")
print(f"Difference (A - B): {a - b}")
print(f"Symmetric Diff (A ^ B): {a ^ b}")

# Membership testing
print(f"\nIs 3 in A? {3 in a}")
print(f"Is 10 in A? {10 in a}")

# Set comprehension
evens = {x for x in range(20) if x % 2 == 0}
print(f"\nEven set: {evens}")
````

### 🆚 Quick Comparison

````python
# Summary table
print("FEATURE        |  LIST    |  TUPLE   |  SET")
print("-" * 50)
print("Ordered        |   ✅     |   ✅     |   ❌")
print("Mutable        |   ✅     |   ❌     |   ✅")
print("Duplicates     |   ✅     |   ✅     |   ❌")
print("Indexed        |   ✅     |   ✅     |   ❌")
print("Hashable       |   ❌     |   ✅     |   ❌")

# When to use what?
print("\n💡 When to use:")
print("• List  → Ordered collection that changes often")
print("• Tuple → Fixed data, faster, dict keys")
print("• Set   → Unique items, membership testing, set math")
````
