**Solution 2 — Filter for Female customers**

**Solution 3 — Average purchase amount by category**

```python
df.groupby('Category')['Purchase Amount (USD)'].mean()
```

**Solution 4 — Bar chart of purchases per category**

```python
category_counts = df['Category'].value_counts()

plt.bar(category_counts.index, category_counts.values, color='steelblue')
plt.title('Purchases by Category')
plt.xlabel('Category')
plt.ylabel('Count')
plt.show()
```

**Solution 5 — Histogram of ages**

```python
plt.hist(df['Age'], bins=10, color='coral', edgecolor='black')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Number of Customers')
plt.show()
```

**Solution 6 — Scatter plot of age vs. purchase amount**

```python
plt.scatter(df['Age'], df['Purchase Amount (USD)'], alpha=0.4, color='seagreen')
plt.title('Age vs Purchase Amount')
plt.xlabel('Age')
plt.ylabel('Purchase Amount (USD)')
plt.show()
```
