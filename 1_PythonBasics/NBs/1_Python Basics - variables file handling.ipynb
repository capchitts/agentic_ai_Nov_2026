## Names and Values in Python

A variable is a name that refers to an object. The object has a value and a type; assignment binds a name to that object. Assigning to the name again makes it refer to the new value.

### Example: Assign a Value

```python
n = 300
print(n)
```

The name `n` now refers to the integer value `300`.

After `n = 300`, the name `n` refers to the integer object whose value is `300`. The next cells show what happens when another name is assigned to the same value.

When Python runs `m = n`, it makes `m` refer to the same object currently referred to by `n`. For the integer example, both names refer to the value `300`.

Reassigning one name later does not reassign the other.

````python
n = 300
print(n)
````

````python
m = n
print(m)
````

**What happens when a name is reassigned?**

Run these statements in order:

```python
n = 300
m = n
m = 400
n = "foo"
```

A variable name refers to a value. Assignment changes which value a name refers to; it does not edit the old value.

| After this statement | `n` refers to | `m` refers to |
| --- | --- | --- |
| `n = 300` | `300` | not assigned yet |
| `m = n` | `300` | `300` |
| `m = 400` | `300` | `400` |
| `n = "foo"` | `"foo"` | `400` |

After the final line, neither `n` nor `m` refers to `300`. If no other part of the program refers to that value, it is no longer reachable by the program. Python manages reclaiming memory for unreachable objects; you do not need to free it yourself.

````python
m = 400
n="foo"
print(m)
print(n)
````

## Making Decisions with `if`, `elif`, and `else`

Python checks conditions from top to bottom. It runs the indented block under the first condition that is true. If none are true, it runs the optional `else` block.

```python
if condition:
    # Runs when condition is True
    statement
elif another_condition:
    # Runs when the first condition was False and this one is True
    statement
else:
    # Runs when all tested conditions were False
    statement
```

Indentation is part of Python syntax: it marks which statements belong to each branch. Use consistent indentation, usually four spaces.

````python
if 1 < 2:
    print('Yep!')
````

### The `else` Branch

The `else` block runs only when every condition above it was false.

````python
if 1 < 2:
    print('first')
else:
    print('last')
````

# File Handling

File handling lets a program read data from files and write data that remains available after the program ends. The examples below cover opening a file, choosing a mode, reading or writing, and closing it safely.

````python
# To open a file we can use open() function, which requires file path and mode as arguments:
````

````python
# Create a small file first, then read it back.
with open("geeks.txt", "w", encoding="utf-8") as file:
    file.write("Hello from Python file handling.\n")

with open("geeks.txt", "r", encoding="utf-8") as file:
    content = file.read()

print(content)
````

## File Modes

The mode tells `open()` what the program intends to do. Choose it carefully, especially when writing because `w` replaces existing contents.

| Mode | Purpose | Important behavior |
| --- | --- | --- |
| `r` | Read text | File must already exist. |
| `r+` | Read and write text | File must exist; existing contents are not automatically erased. |
| `w` | Write text | Creates a file or replaces its contents. |
| `w+` | Write and read text | Creates or replaces the file. |
| `a` | Append text | Adds at the end; creates the file if needed. |
| `a+` | Append and read text | Creates if needed; writes at the end. |
| `x` | Create text file exclusively | Fails if a file with that name already exists. |
| `x+` | Create and read/write exclusively | Fails if the file already exists. |

Add `b` to a mode (for example, `rb` or `wb`) to work with bytes instead of decoded text. `+` enables both reading and writing.

### Closing a File

Closing a file releases its resources and ensures pending writes are completed. Prefer `with open(...)` so Python closes the file automatically when the block ends.

````python
file = open("niti.txt", "r")
# Perform file operations
file.close()
````

### Use a Context Manager

The `with` statement closes the file automatically when the indented block ends, even if an error occurs inside it. This is the recommended pattern for opening files.

````python
with open("geeks.txt", "r") as file:
    content = file.read()
    print(content)
````

````python
# Handling Exceptions When Closing a File
# It’s important to handle exceptions to ensure that files are closed properly, even if an error occurs during file operations.

try:
    file = open("geeks.txt", "r")
    content = file.read()
    print(content)
finally:
    file.close()
````

## File Handling: Benefits and Trade-Offs

| Benefits | Things to consider |
| --- | --- |
| Read and write text, binary, CSV, and other file formats. | Paths, permissions, and missing files can cause errors. |
| Persist data after a program ends. | Choose modes carefully; `w` replaces existing contents. |
| Use the same basic file APIs across operating systems. | Large files and complex formats may need more deliberate processing. |
| Use `with` to close resources reliably. | Never trust unvalidated input to select sensitive file paths. |

# Optional Extension: Task Tracker CLI

This walkthrough combines functions, collections, loops, user input, and file handling in a small command-line project. Read each section, then follow along in a Python file or notebook code cell.

## 1. Import Modules

```python
import json
import os
```

- `json` reads and writes structured task data.
- `os.path.exists()` checks whether the data file already exists.

## 2. Choose a Data File

```python
TASK_FILE = "tasks.json"
```

This constant stores the file path used by the rest of the program.

## 3. Load Existing Tasks

```python
def load_tasks():
    if os.path.exists(TASK_FILE):
        with open(TASK_FILE, "r") as file:
            return json.load(file)
    return []
```

If the file exists, load and return its task list. Otherwise, start with an empty list.

## 4. Save Tasks

```python
def save_tasks(tasks):
    with open(TASK_FILE, "w") as file:
        json.dump(tasks, file, indent=4)
```

`json.dump()` writes the list in JSON format. The `with` block closes the file automatically.

## 5. Add a Task

```python
def add_task(tasks):
    title = input("Enter task title: ")
    tasks.append({"title": title, "done": False})
    print("Task added.")
```

Each task is a dictionary with a title and a completion flag.

## 6. Display Tasks

```python
def view_tasks(tasks):
    if not tasks:
        print("No tasks.")
        return

    for index, task in enumerate(tasks, start=1):
        status = "Done" if task["done"] else "Not done"
        print(f"{index}. [{status}] {task['title']}")
```

`enumerate(..., start=1)` displays human-friendly task numbers while Python still indexes the list from zero.

## 7. Mark a Task Complete

```python
def mark_done(tasks):
    view_tasks(tasks)
    try:
        index = int(input("Task number to mark complete: ")) - 1
        tasks[index]["done"] = True
        print("Task marked complete.")
    except (IndexError, ValueError):
        print("Invalid selection.")
```

The input is converted to an integer, then reduced by one to match Python's zero-based indexing. `ValueError` catches non-numeric input; `IndexError` catches an index outside the list.

## 8. Delete a Task

```python
def delete_task(tasks):
    view_tasks(tasks)
    try:
        index = int(input("Task number to delete: ")) - 1
        tasks.pop(index)
        print("Task deleted.")
    except (IndexError, ValueError):
        print("Invalid selection.")
```

`pop(index)` removes and returns the selected task.

## 9. Build the Menu

```python
def main():
    tasks = load_tasks()

    while True:
        print("\n--- Task Tracker ---")
        print("1. Add task\n2. View tasks\n3. Mark complete\n4. Delete task\n5. Exit")
        choice = input("Choose an option: ")

        if choice == "1":
            add_task(tasks)
        elif choice == "2":
            view_tasks(tasks)
        elif choice == "3":
            mark_done(tasks)
        elif choice == "4":
            delete_task(tasks)
        elif choice == "5":
            save_tasks(tasks)
            print("Goodbye.")
            break
        else:
            print("Choose an option from 1 to 5.")
```

The `while` loop keeps the menu available. Each branch calls the function for that choice; `break` exits when the user selects Exit.

## 10. Run the Program

```python
if __name__ == "__main__":
    main()
```

This guard calls `main()` when the Python file is run directly.
