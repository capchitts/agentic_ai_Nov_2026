## 1. `def` function - defining your own logic in Python

```python
# Syntax of a function (this is just a template, not runnable)

# def function_name(parameters):      # 'def' keyword + name + inputs
#     # code block, indented          # body runs only when function is called
#     return result                   # 'return' sends a value back (optional)
```

```python
# A simple function with NO parameters

def my_first_function():              # define the function name
    print("Hi, this is my first function")   # body: what it does

my_first_function()                   # call the function to run it
```

```python
# A function that adds two numbers

def add(x, y):                        # x and y are parameters (inputs)
    sum_result = x + y                # do the calculation
    return sum_result                 # send the answer back

result = add(2, 5)                    # 2 and 5 are arguments (actual values)
print(result)                         # 7
```

```python
# A function that triples a number

def triple_number(num):               # one input
    tripled = num * 3                 # multiply by 3
    print(tripled)                    # print instead of return

triple_number(5)                      # 15
```

### More simple functions

```python
# Square of a number

def square(n):                        # n = number we want to square
    return n * n                      # return n multiplied by itself

print(square(4))                      # 16
```

```python
# Convert Celsius to Fahrenheit

def to_fahrenheit(celsius):           # input in Celsius
    return (celsius * 9/5) + 32       # standard formula

print(to_fahrenheit(37))              # 98.6
```

```python
# Find the bigger of two numbers

def bigger(a, b):                     # two inputs
    if a > b:                         # check condition
        return a                      # a is bigger
    else:
        return b                      # otherwise b is bigger

print(bigger(10, 25))                 # 25
```

```python
# An average needs at least one value; make the empty-input behavior explicit.
def average(numbers):
    if not numbers:
        raise ValueError("average() requires at least one number")
    return sum(numbers) / len(numbers)


print(average([10, 20, 30, 40]))  # 25.0
```

```python
# Count vowels in a word

def count_vowels(word):               # word = a string
    count = 0                         # start counter at 0
    for ch in word.lower():           # go through each letter (lowercase)
        if ch in "aeiou":             # is it a vowel?
            count = count + 1         # yes -> increase counter
    return count                      # give back the total

print(count_vowels("DataMantra"))     # 4
```

```python
# Reverse a string

def reverse_text(text):               # text = a string
    return text[::-1]                 # slicing with step -1 reverses it

print(reverse_text("python"))         # nohtyp
```

### Default parameter values

```python
# 'exponent' has a DEFAULT value of 2

def power(base, exponent=2):          # if exponent not given, 2 is used
    return base ** exponent           # base raised to exponent

print(power(3))                       # 9  -> uses default 2
print(power(3, 4))                    # 81 -> uses given 4
```

```python
# Greeting with a default greeting word

def greetings(name, greeting="Hello"):   # greeting defaults to 'Hello'
    print(greeting, '! ', name, sep='')  # sep='' removes extra spaces

greetings('John')                        # Hello! John
greetings('John', 'Hi')                  # Hi! John
```

### Returning more than one value

```python
# A function can return multiple values as a tuple

def add_and_subtract(num1, num2):
    sum_result = num1 + num2          # addition
    difference_result = num1 - num2   # subtraction
    return sum_result, difference_result   # return BOTH values

addition_result, subtraction_result = add_and_subtract(8, 3)   # unpack them

print(f"Sum: {addition_result}")            # Sum: 11
print(f"Difference: {subtraction_result}")  # Difference: 5
```

## 2. `*args` - any number of positional arguments

```python
# *args collects all extra values into a TUPLE

def calculate_sum(*args):             # * means 'take as many as you want'
    print(args)                       # see what args looks like
    return sum(args)                  # add them all up

print(calculate_sum(2, 3, 4, 5))      # (2, 3, 4, 5) then 14
print(calculate_sum(4, 5, 6, 7))      # (4, 5, 6, 7) then 22
```

```python
# args works with ZERO arguments too

def total(*args):
    return sum(args)                  # sum of empty tuple = 0

print(total())                        # 0
print(total(10))                      # 10
print(total(10, 20, 30))              # 60
```

```python
# Print every name passed in

def show_names(*names):               # names = tuple of all inputs
    for n in names:                   # loop through them one by one
        print("Hello", n)

show_names("Tarun", "Nathan", "John")
```

```python
# Mixing a normal parameter with *args

def shopping_bill(customer, *prices):   # customer is fixed, prices is flexible
    print("Customer:", customer)
    print("Total:", sum(prices))        # add up all the prices

shopping_bill("Riya", 100, 250, 60)     # Customer: Riya / Total: 410
```

```python
# *nums is a tuple; require at least one value before calling max().
def biggest(*nums):
    if not nums:
        raise ValueError("biggest() requires at least one number")
    return max(nums)


print(biggest(3, 9, 1))
print(biggest(5, 2, 88, 40, 7))
```

## 3. `**kwargs` - any number of keyword (named) arguments

```python
# **kwargs collects named values into a DICTIONARY

def show_details(**kwargs):           # ** means 'take any name=value pairs'
    print(kwargs)                     # see what kwargs looks like
    for key, value in kwargs.items():   # loop over key/value pairs
        print(key, ":", value)

show_details(name="Tarun", city="Delhi", role="Analyst")
```

```python
# A profile function using kwargs

def profile(**info):                  # info is a dictionary
    print("Name :", info.get("name", "Not given"))   # .get() avoids errors
    print("Age  :", info.get("age", "Not given"))    # default if key missing

profile(name="Nathan", age=30)
profile(name="John")                  # age will show 'Not given'
```

```python
# Normal + *args + **kwargs together (order matters)

def order(item, *sizes, **extras):    # fixed, then *args, then **kwargs
    print("Item  :", item)            # single value
    print("Sizes :", sizes)           # tuple
    print("Extras:", extras)          # dictionary

order("Pizza", "Medium", "Large", cheese=True, spicy=False)
```

```python
# Add all numbers passed by name

def add_scores(**subjects):           # e.g. maths=90, english=80
    return sum(subjects.values())     # .values() gives just the numbers

print(add_scores(maths=90, english=80, science=75))   # 245
```

## 4. A slightly bigger example

```python
# Recursive factorial for non-negative integers.
def factorial(n):
    if n < 0:
        raise ValueError("factorial is not defined here for negative integers")
    if n in (0, 1):
        return 1
    return n * factorial(n - 1)


print(factorial(5))  # 120
```

```python
# Check if a number is odd or even

def odd_even(x):
    if x % 2 == 0:                    # % gives the remainder
        print('Its Even')             # remainder 0 -> even
    else:
        print('Its odd')              # otherwise odd

odd_even(6)
odd_even(7)
```

```python
# Recursive function - a function that calls itself (factorial)

def factorial(n):
    if n == 0 or n == 1:              # base case: stops the recursion
        return 1
    else:
        return n * factorial(n - 1)   # calls itself with a smaller number

print(factorial(5))                   # 120
```

## 5. `lambda` function

A **lambda** is a small, one-line, *anonymous* function (a function without a name).

`lambda arguments: expression`

- **lambda** : the keyword
- **arguments** : the inputs
- **expression** : the single operation; its result is returned automatically

Best used for **small, one-time** operations.

```python
# A lambda that triples a number

triple = lambda x: x * 3              # x is input, x*3 is returned

print(type(triple))                   # <class 'function'>
print(triple(5))                      # 15
```

```python
# Same thing written as a def, for comparison

def triple_def(x):                    # normal function
    return x * 3

print(triple_def(5))                  # 15 - same result, more lines
```

```python
# Square using lambda

square = lambda x: x * x              # one input
print(square(5))                      # 25
```

```python
# Addition with two arguments

add = lambda a, b: a + b              # two inputs
print(add(3, 4))                      # 7
```

```python
# Multiplication with lambda

multiply = lambda a, b: a * b
print(multiply(3, 5))                 # 15
```

```python
# Lambda with three arguments

x = lambda a, b, c: a + b + c         # as many inputs as you like
print(x(5, 6, 2))                     # 13
```

```python
# Check even number with lambda

is_even = lambda x: x % 2 == 0        # returns True or False

print(is_even(4))                     # True
print(is_even(5))                     # False
```

```python
# Condition inside a lambda (if-else in one line)

max_number = lambda x, y: x if x > y else y   # return x if bigger, else y

print(max_number(5, 7))               # 7
print(max_number(10, 3))              # 10
```

```python
# Label a value as Profitable / Un-Profitable

conditional = lambda n: "Profitable" if n > 2 else "Un-Profitable"

print(conditional(5))                 # Profitable
print(conditional(1))                 # Un-Profitable
```

```python
# A few more tiny lambdas

to_upper = lambda text: text.upper()          # make text uppercase
add_gst  = lambda price: price * 1.18         # add 18% GST
is_adult = lambda age: age >= 18              # True / False

print(to_upper("datamantra"))                 # DATAMANTRA
print(add_gst(100))                           # 118.0
print(is_adult(15))                           # False
```

### Where lambda is really used

Mostly with functions like `filter()`, `sorted()` and with **pandas** `.apply()`.

```python
# lambda with filter() - keep only values greater than 10

numbers = [4, 1, 7, 3, 6, 12, 18, 16, 50, 10, 4, 5]

numbers_gt_10 = list(filter(lambda x: x > 10, numbers))   # filter keeps True ones

print(numbers_gt_10)                  # [12, 18, 16, 50]
```

```python
# lambda with sorted() - sort names by their length

names = ["Tarun", "Jo", "Nathan", "Ria"]

print(sorted(names, key=lambda n: len(n)))    # ['Jo', 'Ria', 'Tarun', 'Nathan']
```

```python
# lambda inside pandas

import pandas as pd

df = pd.DataFrame({'A': [1, 2, 3]})           # small dataframe

df['B'] = df['A'].apply(lambda x: x * 10)     # apply lambda to every row

df.head(3)
```

## Quick summary

| Concept | Meaning |
|---|---|
| `def` | Creates a named, reusable function |
| `return` | Sends a value back to the caller |
| default value | Used when the argument is not passed |
| `*args` | Any number of positional arguments -> **tuple** |
| `**kwargs` | Any number of named arguments -> **dictionary** |
| `lambda` | Small one-line anonymous function |

## Function Checkpoint

A reusable function has a clear name, inputs, a body, and a defined result. Parameters are names in the definition; arguments are values at the call. Use `return` when later code needs the result; use `print` only to display it. Defaults cover omitted inputs, `*args` gathers positional values into a tuple, and `**kwargs` gathers named values into a dictionary.
